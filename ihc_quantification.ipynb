```python
#data /content/drive/MyDrive/ppgl datasets/IHC pheos SOX2 quantification.xlsx
data = pd.read_excel('/content/drive/MyDrive/ppgl datasets/IHC pheos SOX2 quantification.xlsx',header=None)
#rename to sample and percentage
data.columns = ['sample','percentage']
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(2.5, 4))
sns.boxplot(data=data, y='percentage', ax=ax, color='lightblue')
#vline at mean and legend
ax.axhline(data['percentage'].mean(), color='green', linestyle='--', label='Mean')
sns.stripplot(data=data, y='percentage', color='black', ax=ax, size=8,alpha=0.4)
ax.set_title('Percentage of SOX2+ cells across PPGL samples')
ax.set_xlabel('', fontsize=12)
ax.set_ylabel('Percentage', fontsize=12)
#ylim 0 -10
ax.set_ylim(0,10)
#remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
#larger ticks x
ax.xaxis.set_tick_params(labelsize=12)
#add legend for mean line
plt.tight_layout()
#save as svg and png
plt.savefig('boxplot.svg', format='svg')
plt.savefig('boxplot.png', format='png')
plt.show()
```
